One-Hot Encoding  

One-Hot Encoding is a method of converting categorical variables into multiple binary columns, where only the matching category is marked as 1 and all other categories are marked as 0.

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
df = pd.DataFrame({
    'color': ['red', 'blue', 'green', 'green', 'red', 'blue']
})
```

```python
encoder=OneHotEncoder()

print("Original DataFrame:")
print(df)
```

```python
# Perform One-Hot Encoding
encoded = encoder.fit_transform(df[['color']]).toarray()
```

```python
encoder_df = pd.DataFrame(
    encoded,
    columns=encoder.get_feature_names_out()
)
```

```python
print("\nEncoded DataFrame:")
print(encoder_df)
```
